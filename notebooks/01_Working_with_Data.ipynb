# Resampling & Creating Regular Time Grids

| Alias | Frequency | Example |
|:-----:|:---------:|:--------|
| `s` | Secondly | `'10s'` (every 10 seconds) |
| `min` | Minutely | `'5min'` (every 5 minutes) |
| `h` | Hourly | `'3h'` (every 3 hours) |
| `D` | Calendar day | `'D'` or `'2D'` |
| `B` | Business day | `'B'` (skips weekends) |
| `W` | Weekly | `'W'` (defaults to Sunday) |
| `ME` | Month End | `'ME'` (last day of calendar month) |
| `MS` | Month Start | `'MS'` (first day of calendar month) |
| `QE` | Quarter End | `'QE'` (last day of the quarter) |
| `QS` | Quarter Start | `'QS'` (first day of the quarter) |
| `YE` | Year End | `'YE'` (last day of calendar year) |
| `YS` | Year Start | `'YS'` (first day of calendar year) |

```python
flow_data.resample("30min").mean().head(5)
```

```python
flow_data.resample("D").mean().head(5)
```

```python
flow_data.resample("2h").min().head(5)
```

# Multiple Aggregations

```python
flow_data.resample("D").agg(["mean", "max", "min", "std", "count"]).head(5)
```

```python
flow_data.loc[:, "dma_3"].resample("D").agg(["mean", "max", "min", "std", "count"]).head(5)
```

# Chaining Commands

```python
df3 = (
    flow_data.loc[:, ["dma_3"]]
    .resample("D")
    .mean()
    .rename(columns={"dma_3": "dma_3_daily_mean"})
    .loc["2021-01-01":"2021-01-10"]
)
df3.head(5)
```

# Merging Multiple Datasets

```python
merged_data = pd.merge_asof(
    flow_data.sort_index(),
    weather_data.sort_index(),
    left_index=True,
    right_index=True,
    direction="nearest",
)
merged_data.head(5)
```
